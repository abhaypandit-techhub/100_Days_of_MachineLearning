#### Skewness measures how much a data distribution is asymmetric or tilted to one side.
how much tiled data distributed in left and right

| Value | Meaning                                   |
| ----- | ----------------------------------------- |
| `≈ 0` | Symmetric distribution                    |
| `> 0` | **Right/positive skew** → tail right side |
| `< 0` | **Left/negative skew** → tail left side   |


```python
df['Age'].skew()
```
