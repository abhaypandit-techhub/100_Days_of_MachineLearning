```python
date.info()
```

## 1 Extract year

```python
# date of year
date['date_year']=date['date'].dt.year
date.head(2)
```

```python
# day of month number
date['date_month_no.']=date['date'].dt.month
date.head(2)
```

### If you need month

```python
# day of month with name
date['date_month_name']=date['date'].dt.month_name()
date.head(2)
```

## Extract Days

```python
# day of week
date['date_day']=date['date'].dt.day
date.head(2)
```

```python
# day with name
date['date_day_name']=date['date'].dt.day_name()
date.head(2)
```

```python
# day of week
date['date_dow']=date['date'].dt.dayofweek
date.head(2)
```

```python
# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()
date.head(2)
```

## Is weekend ?

```python
date['date_is_weekend']=np.where(date['date_dow_name'].isin(['Sunday','Saturday']),1,0)
date.head(2)
```

## Extract week of the year

```python
# date['date_week']=date['date'].dt.week  # This is old version removed from the pandas new version
date['date_week'] = date['date'].dt.isocalendar().week
date.head(2)
```

## Extract Quater

```python

date['quarter'] = date['date'].dt.quarter
date.head(2)
```
