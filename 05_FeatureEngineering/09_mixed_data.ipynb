```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day33-handling-mixed-variables/titanic.csv")
```

```python
df.head()
```

```python
df['number'].unique() # A is stand for alone
```

```python
fig=df['number'].value_counts().plot(kind='bar') 
fig.set_title("passengers travelling with:")
```

```python
# extract numerical part
df['number_numerical']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
df.head(10)
```

```python
# extract categorical part
df['number_categorical'] = np.where(
    df['number_numerical'].isnull(),
    df['number'],
    np.nan
)
```

```python
df.head()
```

```python
df['Cabin'].unique()
```

\d → any digit from 0 to 1
+ → one or more than one

```python
# capture numeric part
df['cabin_num']=df['Cabin'].str.extract('(\d+)') # \d+ any digit 0 to 9

# df['cabin_cat']=df['Cabin'].str[0] # capture the first letter 

df['cabin_cat']=df['Cabin'].str.extract('([A-Za-z]+)') # capture the letter whether at the begnning ,mid and end all letter(+)
```

```python
df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

condition              True                    False
   ↓                    ↓                      ↓
isdigit()?             NaN                  original value 

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])
```
