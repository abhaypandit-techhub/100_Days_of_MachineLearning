```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(
    "https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day44-outlier-detection-using-percentiles/weight-height.csv"
)
```

```python
df.head()
```

```python
df.shape
```

```python
sns.distplot(df['Height'])
```

```python
sns.boxplot(df['Height'],orient='h')
plt.grid()
```

```python
upper_limit = df['Height'].quantile(0.99)
upper_limit
```

```python
lower_limit = df['Height'].quantile(0.01)
lower_limit
```

```python
df[(df['Height'] >= upper_limit) | (df['Height'] <= lower_limit)]
```

```python
df['Height'].describe()
```

## Trimming

```python
new_df = df[(df['Height'] <= upper_limit) & (df['Height'] >= lower_limit)]
```

```python
new_df['Height'].describe()
```

```python
sns.displot(new_df['Height'],kde=True)
```

```python
sns.boxplot(new_df['Height'],orient='h')
```

### Capping ( Winsorization )

```python
df["Height"] =np.where(df['Height'] >= upper_limit , upper_limit , np.where ( df['Height'] <= lower_limit , lower_limit ,df['Height']))
```
