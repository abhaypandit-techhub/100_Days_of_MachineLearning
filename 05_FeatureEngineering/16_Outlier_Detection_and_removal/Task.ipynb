```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("house_price_outliers.csv")
```

```python
df.shape
```

```python
df.head()
```

```python

plt.figure(figsize=(16,5))

plt.subplot(1,2,1)
sns.histplot(df['age_years'], kde=True)
plt.title("Age Distribution")

plt.subplot(1,2,2)
sns.histplot(df['distance_to_center_km'], kde=True)
plt.title("Distance Distribution")

plt.tight_layout()
plt.show()
```

```python
lower_limit = df['age_years'].mean() - 3*df['age_years'].std()
upper_limit = df['age_years'].mean() + 3*df['age_years'].std()
```

```python
upper_limit
lower_limit
```

```python
new_df = df[(df['age_years'] > lower_limit) &  (df['age_years'] < upper_limit) ]
```

```python

plt.figure(figsize=(16,5))

plt.subplot(1,2,1)
sns.histplot(df['age_years'], kde=True)
plt.title("Age Distribution")

plt.subplot(1,2,2)
sns.histplot(new_df['age_years'], kde=True)
plt.title("Age Distribution")

plt.tight_layout()
plt.show()
```
